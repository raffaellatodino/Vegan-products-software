# Software di un negozio di prodotti vegani

### Import pacchetti

```python
import json
```

### Definizione delle funzioni necessarie

```python
file_name = "inventory_market.json"


def adding_new_product(inventory, name, quant, sale_price, purch_price):

    '''
    Apre o crea un file json con una lista di dizionari che rappresenta un magazzino e ci aggiunge un nuovo prodotto.
    Ogni dizionario della lista rappresenta un prodotto.

    Args:
        inventory (list): lista che rappresenta il magazzino
        name: nome del prodotto da aggiungere a magazzino
        quant (int): quantità di prodotto da aggiungere
        sale_price (float): prezzo di vendita del prodotto
        purch_price (float): prezzo di acquisto del prodotto

    '''

    with open(file_name, 'w+') as json_file:
        inventory.append({"PRODOTTO": name,
                        "QUANTITA": quant,
                        "PREZZO_VENDITA": sale_price,
                        "PREZZO_ACQ": purch_price})
        json.dump(inventory, json_file, indent=3)



def adding_existing_product(inventory, name, quant):

    '''
    Apre il file del magazzino e aggiunge la quantità specificata ad un prodotto esistente.

    Args:
        inventory (list): magazzino
        name: nome del prodotto
        quant (int): quantità da aggiungere

    '''

    with open(file_name, 'r+') as json_file:
        for el in inventory:

            if el['PRODOTTO'] == name:
                el["QUANTITA"] += quant

                json_file.seek(0)
                json_file.truncate()
                json.dump(inventory, json_file, indent=3)



def create_list(inventory):

    ''''
    Crea una lista dei prodotti a magazzino, togliendo il prezzo di acquisto e rinominando il prezzo di vendita come "prezzo".

    Args:
        inventory (list): magazzino
    '''

    for el in inventory:
        el['PREZZO'] = el.pop('PREZZO_VENDITA')
        el.pop('PREZZO_ACQ')



def print_product_list():

    '''
    Stampa l'elenco dei prodotti presenti in magazzino

    '''

    with open(file_name) as json_file:
        inv = json.load(json_file)
        elenco = inv.copy()
        create_list(elenco)
        print("%s\t%s\t%s" % tuple(elenco[0].keys()))
        for el in inv:
            print("%s\t\t%i\t\t%s%.2f" % (el['PRODOTTO'], el["QUANTITA"], '€',el['PREZZO']))



def sale(inventory, sales, name, quant):

    '''
    Rimuove dal magazzino i prodotti venduti e stampa la vendita effettuata.

    Args:
        inventory (list): magazzino
        sales (list): lista inizialmente vuota nella quale vengono registrate le vendite
        name: nome del prodotto da vendere
        quant: quantità del prodotto da vendere

    '''

    with open(file_name, 'r+') as json_file:
        for el in inventory:

            if el['PRODOTTO'] == name:
                try:
                    assert (quant <= el["QUANTITA"]), 'La quantità indicata non è a magazzino'
                    el["QUANTITA"] -= quant

                except AssertionError as e:
                    print(e)

                json_file.seek(0)
                json_file.truncate()
                json.dump(inventory, json_file, indent=3)

    sales.append({"PRODOTTO": name,
                        "QUANTITA": quant})

    for el in sales:
        for i in inventory:
            if el['PRODOTTO'] == i['PRODOTTO']:
                el['guadagno_lordo'] = el["QUANTITA"]* i["PREZZO_VENDITA"]
                el['guadagno_netto'] = el["QUANTITA"]* (i["PREZZO_VENDITA"] - i["PREZZO_ACQ"])



def aiuto():

    '''
    Stampa i comandi inseribili nel programma

    '''

    print("I comandi possibili sono i seguenti:\n" \
    "aggiungi: aggiungi un prodotto al magazzino\n" \
    "elenca: elenca i prodotti in magazzino\n" \
    "vendita: registra una vendita effettuata\n" \
    "profitti: mostra i profitti totali\n" \
    "aiuto: mostra i possibili comandi\n" \
    "chiudi: esci dal programma")



def name_exists(list):

    '''
    Consente di inserire il nome di un prodotto e verifica che sia presente in una lista (es. nel magazzino).
    Chiede il nome del prodotto fino ache non ne viene inserito uno presente.

    Args:
        list (list): lista in cui verificare se il nome del prodotto è presente.

    '''

    name_exists = False
    while not name_exists:
      try:
        name = input('Nome del prodotto: ')
        assert(name in list), 'Prodotto non presente in magazzino'

        name_exists = True
        return name
      except AssertionError as e:
        print(e)



def assert_int():

    '''
    Consente di inserire un numero e verifica che sia intero. Se viene inserito un float, se la cifra decimale è zero lo converte in intero, altrimenti scatena un'eccezione.
    Chiede di inserire un numero fino a che non viene inserito un valore accettabile.
    '''

    is_float = False
    while not is_float:
        try:
            quant = float(input("Quantità: "))
            assert(quant == int(quant)), "Inserisci un numero intero!"
            quant = int(quant)
            is_float = True
            return quant
        except ValueError:
            print(f"Inserisci un valore numerico!")
        except AssertionError as e:
            print(e)



def is_quant_present(inventory, name):

    '''
    Consente di inserire la quantità di un prodotto a magazzino e verifica che sia presente.
    Verifica che la quantità inserita sia un numero intero e che sia presente in magazzino, altrimenti scatena un'eccezione.
    Chiede di inserire la quantità fino a che non ne viene inserita una presente.

    Args:
        inventory (list): magazzino in cui verificare
        name: nome del prodotto di cui verificare la quantità

    '''

    is_quant_present = False
    while not is_quant_present:
      for el in inventory:
        if el['PRODOTTO'] == name:
          try:
            quant = assert_int()
            assert(quant <= el["QUANTITA"]), "La quantità indicata non è a magazzino"

            is_quant_present = True
            return quant
          except AssertionError as e:
             print(e)



def is_float(type):
    '''
    Consente di inserire un numero e verifica che sia un float. Se viene inserito un intero lo converte in float, se viene inserita una stringa scatena un'eccezione.

    Args:
    type (string): tipologia di numero da inserire. Può essere 'vendita','acquisto' o 'generico'.

    '''

    is_float = False
    while not is_float:
        try:
            if type == 'generico':
                number = float(input("Quantità: "))
            else:
                number = float(input(f"Prezzo di {type}: "))
            is_float = True
            return number
        except ValueError:
            print('Inserisci un valore numerico!')
```

### Codice del programma

```python
cmd = None

#lista necessaria per registrare le vendite totali e stamparle con il comando "profitti"
total_sales = []


while cmd!="chiudi":

  #apre il file json del magazzino, se non esiste lo crea
  try:
      with open(file_name) as json_file:
          inventory = json.load(json_file)
  except FileNotFoundError:
      inventory = []

  #tiene in magazzino solo i prodotti con quantità positiva
  inventory = [el for el in inventory if el["QUANTITA"] > 0]
  with open(file_name, 'w') as json_file:
    json.dump(inventory, json_file, indent=3)



  cmd = input("Inserisci un comando: ")


  if cmd=="aggiungi":
    # aggiunge un prodotto al magazzino, se è già presente ne aggiunge la quantità

    product_name = input("Nome del prodotto: ")
    product_quant = assert_int()

    products = [el['PRODOTTO'] for el in inventory]

    if product_name in products:
        adding_existing_product(inventory,product_name,product_quant)
        print(f"AGGIUNTO: {product_quant} x {product_name}")

    elif product_name not in products:

        purchase_price = is_float('acquisto')
        sale_price = is_float('vendita')

        adding_new_product(inventory, product_name,product_quant,sale_price,purchase_price)
        print(f"AGGIUNTO: {product_quant} x {product_name}")



  elif cmd=="elenca":
    # elenca tutti i prodotti nel magazzino

    try:
      print_product_list()
    except FileNotFoundError:
       print('Nessun prodotto presente')
    except IndexError:
       print('Nessun prodotto presente')



  elif cmd=="vendita":
    # registra una vendita

    products = [el['PRODOTTO'] for el in inventory]

    #lista vuota in cui vengono registrate le vendite
    sales = []

    product_name = name_exists(products)
    product_quant = is_quant_present(inventory, product_name)

    sale(inventory, sales, product_name, product_quant)
    adding_other = input('Aggiungere un altro prodotto?[sì/no]')


    while adding_other != 'no':

      product_name = name_exists(products)
      product_quant = is_quant_present(inventory, product_name)

      sale(inventory, sales, product_name, product_quant)
      adding_other = input('Aggiungere un altro prodotto?[sì/no]')


    if adding_other == 'no':
      #stampa la vendita e il totale venduto

      print('VENDITA REGISTRATA')
      temporary_total = 0
      for el in sales:
        print(f"- {el['QUANTITA']} x {el['PRODOTTO']}: €{round(el['guadagno_lordo'],2)}")

        for i in inventory:
          if el['PRODOTTO'] == i['PRODOTTO']:
            temporary_total += el["QUANTITA"]*i["PREZZO_VENDITA"]
      print(f"Totale: €{round(temporary_total, 2)}")

    #aggiunge le vendite effettuate al totale delle vendite
    total_sales = total_sales + sales


  elif cmd=="profitti":
    # mostra profitti netti e lordi

    profitto_lordo = 0
    profitto_netto = 0

    for el in total_sales:
      profitto_lordo+= el['guadagno_lordo']
      profitto_netto += el['guadagno_netto']
    print(f"Profitto: lordo=€{round(profitto_lordo,2)} netto=€{round(profitto_netto,2)}")



  elif cmd=="aiuto":
    # mostra i possibili comandi
    aiuto()


  elif cmd=="chiudi":
    # saluta e interrompi il programma
    print("Bye bye")

  else:
    # comando non valido
    # mostra messaggio di aiuto
    print("Comando non valido")
    aiuto()
```
